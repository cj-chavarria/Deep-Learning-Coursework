## ¿Qué es el Dropout?

El **dropout** es una técnica de regularización utilizada en redes neuronales para evitar el sobreajuste (overfitting) durante el entrenamiento del modelo. Fue propuesta por **Geoffrey Hinton** y su equipo en 2014, y es ampliamente utilizada en la práctica.

### ¿Cómo funciona el dropout?

Durante el entrenamiento, **dropout** "apaga" aleatoriamente un porcentaje de las neuronas en la red neuronal en cada paso de entrenamiento. Esto significa que, en cada iteración, algunas neuronas no participarán en el proceso de propagación hacia adelante ni en la retropropagación del error.

#### El proceso de dropout:
1. **Durante el entrenamiento**:
   - En cada paso de entrenamiento, se selecciona aleatoriamente un porcentaje de neuronas para "apagar", es decir, estas neuronas no se usan en ese paso de entrenamiento.
   - Este porcentaje se define por un parámetro llamado **tasa de dropout** (por ejemplo, 0.2 o 0.5).
   - Si la tasa de dropout es **0.2**, significa que el 20% de las neuronas se apagan aleatoriamente en cada paso de entrenamiento.

2. **Durante la inferencia (prueba)**:
   - Cuando el modelo ya está entrenado y se utiliza para hacer predicciones (en el conjunto de prueba o en producción), **no se apagan neuronas**.
   - En cambio, los pesos de las neuronas se **escalan** para compensar la diferencia entre el entrenamiento (con dropout) y la inferencia (sin dropout). Por ejemplo, si la tasa de dropout es 0.2, los pesos de las neuronas restantes se multiplican por 0.8 durante la prueba.

### ¿Por qué usar dropout?

El principal objetivo de **dropout** es **prevenir el sobreajuste** (overfitting), que ocurre cuando el modelo se adapta demasiado a los datos de entrenamiento, memorizando patrones específicos que no generalizan bien a nuevos datos. Al aplicar dropout, el modelo se ve forzado a aprender representaciones más generales y robustas.

- **Mejora la generalización**: Al "desactivar" aleatoriamente algunas neuronas, el modelo no depende demasiado de ninguna neurona en particular. Esto obliga a la red a encontrar representaciones más sólidas y generalizables para los datos.
  
- **Reducción de la co-adaptación de neuronas**: Las neuronas en una red neuronal pueden "co-adaptarse" a los datos, es decir, ajustarse demasiado a un patrón específico en los datos de entrenamiento. El dropout ayuda a evitar esto al hacer que cada neurona sea más independiente, promoviendo que cada neurona aprenda características únicas.
