# 0. Librerías básicas

```python
#Librerías básicas de esta notebook
import matplotlib.pyplot as plt
import numpy as np
import cv2 as cv
import time
```

```python
video = cv.VideoCapture('./videos/video_original.mp4')
i = 1
while(video.isOpened()):
    ret, frame_original = video.read()
    if ret:
        curr_frame = frame_original.copy()
        primer_frame = curr_frame if i == 1 else primer_frame
        i+=1
        
        frame_hsv = cv.cvtColor(curr_frame, cv.COLOR_RGB2HSV)
        inferior, superior = np.array([30,20,120]), np.array([100,255,255])
        mascara =  cv.inRange(frame_hsv, inferior, superior)
        mascara = cv.erode(mascara, None, iterations=7)
        mascara = cv.dilate(mascara, None, iterations=15)
        primer_frame_mascara = cv.bitwise_and(primer_frame, primer_frame, mask=mascara)
        curr_frame_mascara = cv.bitwise_and(curr_frame, curr_frame, mask=cv.bitwise_not(mascara))
        
        final_frame = cv.add(curr_frame_mascara, primer_frame_mascara)
        
        cv.imshow('video', final_frame)
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
video.release()
cv.destroyAllWindows()
```
