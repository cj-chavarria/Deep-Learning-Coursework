# 0. Librerías básicas

```python
#Librerías básicas de esta notebook
import matplotlib.pyplot as plt
import numpy as np
import cv2 as cv
```

# 1. Reproducir  el video

```python
video = cv.VideoCapture("videos/video_ejemplo.mp4")
while(video.isOpened()):
    ret, frame = video.read()
    #ret es una  variable booleana, true -> existe imagen , false -> No existe (final de video)
    if ret:
        cv.imshow("video original", frame)  
        
        
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
video.release()
cv.destroyAllWindows()
```

# 2. Cambiar el tamaño y tiempo de reproducción

```python
import time


```

```python
video = cv.VideoCapture("videos/video_ejemplo.mp4")
while(video.isOpened()):
    time.sleep(0.0166666666666)
    ret, frame = video.read()
    #ret es una  variable booleana, true -> existe imagen , false -> No existe (final de video)
    if ret:
        #cv.imshow("video original", frame)  
        img = frame.copy()
        img_r = cv.resize(img, (int(img.shape[1]*0.5), int(img.shape[0]*0.5)), interpolation=cv.INTER_NEAREST)
        
        cv.imshow('vide mitad', img_r)
        
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
video.release()
cv.destroyAllWindows()
```

# 3. Guardar los primeros N frames

```python

video = cv.VideoCapture("videos/video_ejemplo.mp4")
frames_10 = []
i=0
while video.isOpened() and i < 10:
    ret, frame = video.read()
    #ret es una  variable booleana, true -> existe imagen , false -> No existe (final de video)
    if ret:
        cv.imshow("video original", frame)  
        img =  frame.copy()
        cv.imwrite(f'capturas/imagen_{i}.jpg', img)
        i += 1
        
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
video.release()
cv.destroyAllWindows()
```

# 4. Cambiar a escala de grises

```python
#Definir la función 
video = cv.VideoCapture("videos/video_ejemplo.mp4")
while(video.isOpened()):
    time.sleep(0.0025)
    ret, frame = video.read()
    #ret es una  variable booleana, true -> existe imagen , false -> No existe (final de video)
    if ret:
        #cv.imshow("video original", frame)  
        img = frame.copy()
        img_r = cv.resize(img, (int(img.shape[1]*0.5), int(img.shape[0]*0.5)), interpolation=cv.INTER_NEAREST)
        img_gris = cv.cvtColor(img_r, cv.COLOR_BGR2GRAY)
        
        _, img_bin = cv.threshold(img_gris, 200, 255, cv.THRESH_BINARY)
        
        
        cv.imshow('vide mitad', img_r)
        cv.imshow('video + resize + gris', img_gris)
        cv.imshow('video + resize + gris + binarizado', img_bin)
        
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
video.release()
cv.destroyAllWindows()
```
