```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
```

```python
data_train = pd.read_csv('./DLCV_09_ReconocimientoNumeros_ML/datos/mnist_train_tarea.txt')
data_test = pd.read_csv('./DLCV_09_ReconocimientoNumeros_ML/datos/mnist_test_clase.txt')

X_train = data_train.iloc[:, 1:]
y_train = data_train.iloc[:, 0]

X_test = data_test.iloc[:, 1:]
y_test = data_test.iloc[:, 0]
```

```python
steps = [ ('scaler', StandardScaler()), ('SVM', SVC(kernel='poly'))]
SVM_model = Pipeline(steps)
```

```python
params = {
    'SVM__C': [0.001, 0.01, 0.1, 1, 10, 100],
    'SVM__gamma': [0.001, 0.01, 0.1, 1, 10, 100],
    'SVM__degree': [2, 3, 4],
    'SVM__kernel': ['rbf', 'poly', 'sigmoid']
}
```

```python
grid_SVM =  GridSearchCV(SVM_model, param_grid=params, cv=5, n_jobs=-1)
grid_SVM.fit(X_train, y_train)
```
